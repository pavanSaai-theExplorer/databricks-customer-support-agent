```python
# Import required libraries for OpenAI API interaction
from openai import OpenAI
import os

# Authenticate with Databricks - Get the current workspace token
# How to get your Databricks token: https://docs.databricks.com/en/dev-tools/auth/pat.html
# DATABRICKS_TOKEN = os.environ.get('DATABRICKS_TOKEN')  # Alternative: Use environment variable
# For Databricks notebooks, use the built-in context to retrieve the token automatically
DATABRICKS_TOKEN = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()

# Initialize OpenAI client with Databricks serving endpoint
# The base_url points to the Databricks model serving endpoint
client = OpenAI(
    api_key=DATABRICKS_TOKEN,
    base_url="https://dbc-4248f24a-aa8c.cloud.databricks.com/serving-endpoints"
)

# Make a chat completion request to the LLM
# Using databricks-gpt-oss-120b model to answer a question about AI's impact on data engineering
response = client.chat.completions.create(
    model="databricks-gpt-oss-120b",
    messages=[
        {
            "role": "user",
            "content": "Hey, How Data Engineering is being changed because of AI in brief?"
        }
    ],
    max_tokens=5000  # Maximum number of tokens in the response
)

# Display the LLM's response
print(response.choices[0].message.content)
```

```python
# Attempt to extract text from the response content
# Note: This will fail because response.choices[0].message.content is a string, not a list
# The correct way to access the content is: response.choices[0].message.content
print(response.choices[0].message.content[1].get('text'))
```

**AI‑driven shifts in data engineering (in a nutshell)**  

| Area | Traditional approach | AI‑enhanced approach | Impact |
|------|----------------------|----------------------|--------|
| **Data ingestion & integration** | Hand‑crafted connectors, manual schema mapping | *Auto‑discovery* of source schemas, *intelligent adapters* that infer formats (JSON, XML, binary) and generate pipelines on the fly | Faster onboarding of new data sources; less boiler‑plate code |
| **ETL/ELT orchestration** | Scripted jobs (SQL, Python, Spark) that require explicit transformation logic | *LLM‑assisted code generation* and *self‑optimizing pipelines* that rewrite queries, suggest partitioning, and auto‑tune resources | Shorter development cycles; runtime cost reductions |
| **Data quality & profiling** | Rule‑based checks, static data‑validation scripts | *Anomaly‑detection models* that learn normal distributions, detect drift, and surface outliers without explicit thresholds | Early detection of corruption, higher trust in downstream analytics |
| **Metadata & cataloging** | Manual tagging, static documentation | *Embedding‑based metadata extraction* (e.g., column‑level semantics from table samples) and *LLM‑powered search* (“find all tables that contain customer address”) | Better discoverability, smoother data governance |
| **Schema evolution & inference** | Versioned DDL, manual migration scripts | *Probabilistic schema inference* that proposes version updates and auto‑generates migration code | Less downtime, smoother handling of semi‑structured data |
| **Observability & monitoring** | Metric dashboards, alerts on thresholds | *Predictive alerting* (forecasted job failures, resource bottlenecks) and *root‑cause explanation* generated by LLMs | Proactive issue resolution, reduced MTTR |
| **Data lineage & impact analysis** | Static lineage graphs built from ETL scripts | *AI‑reconstructed lineage* from logs, notebooks, and code commits, with “what‑if” impact simulations | Faster impact assessments for schema changes or deprecations |
| **Data preparation for ML** | Separate feature‑engineering pipelines | *Feature stores powered by AI* that auto‑suggest feature definitions, perform drift detection, and suggest enrichment sources | Tightens the ML‑ops loop; reduces duplication of effort |
| **Synthetic & augmentation data** | Manual sampling or rule‑based augmentation | *Generative models* (GANs, diffusion) that create realistic test data, fill data gaps, or simulate edge cases | Improves testing, privacy‑preserving data sharing |
| **Skill set & tooling** | Heavy focus on SQL, Spark, Scala, Airflow | Blend of **data engineering + prompt engineering**; use of LLM‑augmented IDEs, low‑code pipeline builders, and AI‑observability platforms | Engineers spend more time on design & validation, less on rote coding |

### Key take‑aways

1. **Automation at scale** – AI now writes, tunes, and monitors large parts of the pipeline, shrinking development time from weeks to days (or even hours).
2. **Higher data trust** – Continuous, model‑driven quality checks catch anomalies that rule‑based checks miss.
3. **Improved discoverability** – Semantic embeddings and LLM‑powered search turn massive catalogs into searchable knowledge bases.
4. **Shift in skill set** – Engineers need to understand prompting, model evaluation, and AI‑tooling alongside classic SQL/ETL skills.
5. **Governance remains crucial** – AI introduces new risks (bias, hallucination, opaque transformations), so robust oversight, lineage, and testing frameworks are more important than ever.

In short, AI is turning data engineering from a largely manual, code‑centric discipline into a **semi‑automated, intelligent orchestration layer**, letting teams focus on strategy, reliability, and governance rather than repetitive plumbing.

**AI‑driven shifts in Data Engineering (in a nutshell)**  

| Area | Traditional approach | AI‑enhanced approach | What it means for data engineers |
|------|----------------------|----------------------|----------------------------------|
| **Pipeline building & orchestration** | Hand‑coded ETL jobs, manual scheduling (Airflow, Luigi, etc.) | *Auto‑pipeline generation* – LLMs suggest DAGs, write Spark/SQL code, and auto‑tune resources. | Spend less time writing boilerplate; focus on architecture, edge‑case handling, and validation. |
| **Data ingestion & integration** | Fixed connectors, custom scripts for each source. | *Intelligent connectors* that infer schema, map fields, and resolve format mismatches using ML. | Faster onboarding of new sources; fewer “schema‑drift” bugs. |
| **Data quality & testing** | Rule‑based checks, static tests written manually. | *AI‑based anomaly detection* and *automated test generation* (e.g., Great Expectations + LLM‑suggested expectations). | Continuous, adaptive quality monitoring with less manual rule‑crafting. |
| **Metadata & cataloging** | Manual tagging, limited lineage capture. | *Semantic enrichment* – LLMs extract business meanings, auto‑populate tags, and generate natural‑language documentation. | Richer catalogs, easier data discovery, and better hand‑off to analysts. |
| **Data transformation (ELT)** | SQL/Scala/Python scripts written by engineers. | *Code‑completion & transformation synthesis* – LLMs propose optimized Spark/SQL snippets, suggest push‑down predicates, and auto‑vectorize. | Higher productivity, fewer performance pitfalls. |
| **Observability & debugging** | Log‑based alerts, manual root‑cause analysis. | *AI‑driven anomaly attribution* – models correlate metric spikes with code changes, data drift, or infrastructure events. | Faster MTTR (Mean Time to Resolve) and proactive issue prevention. |
| **Synthetic & augmenting data** | Manual data generation scripts. | *Generative models* (e.g., diffusion, GANs) create realistic test data or fill gaps for downstream ML. | Safer sharing of data, richer training sets without extra collection effort. |
| **Governance & compliance** | Rule‑based masking, static policy checks. | *Policy‑as‑code + LLM audit* – AI scans pipelines for PII, suggests masking, and produces compliance reports. | Reduced risk, automated audit trails. |
| **Query & storage optimization** | Cost‑based optimizer, manual tuning. | *Reinforcement‑learning optimizers* that learn workload patterns and auto‑select partitioning, indexing, or materialized views. | Lower cloud spend, better latency without manual tuning. |
| **Collaboration with ML teams (MLOps)** | Separate ETL and model training pipelines. | *Unified data‑ML orchestration* – AI links data versioning (e.g., Delta Lake) with model registries, auto‑triggers retraining on drift. | Seamless hand‑off, tighter feedback loops. |

### Key Take‑aways for Data Engineers
1. **Automation of repetitive work** – LLMs and ML models write, test, and tune code, freeing you for higher‑level design and problem‑solving.  
2. **Smarter observability** – AI surfaces root causes and predicts failures before they surface.  
3. **Richer metadata** – Natural‑language descriptions and semantic tags make data catalogs far more usable.  
4. **New skill set** – Comfort with prompting LLMs, evaluating AI‑generated code, and understanding model‑driven optimizers becomes valuable.  
5. **Governance at scale** – AI helps enforce privacy and compliance automatically, reducing manual policy checks.  

In short, AI is turning data engineering from a largely manual, code‑centric discipline into a **semi‑automated, intelligence‑augmented practice**, where engineers act more as architects, validators, and overseers of AI‑generated pipelines.
